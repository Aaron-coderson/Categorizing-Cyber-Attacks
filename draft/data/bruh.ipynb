```python
import docx
from docx.enum.text import WD_COLOR_INDEX


def extract_headings_and_highlights(input_path, output_path):
    # Open the input document
    doc = docx.Document(input_path)

    # Create a new document for output
    new_doc = docx.Document()

    # Track if we've found any content
    found_content = False

    # Iterate through paragraphs in the input document
    for para in doc.paragraphs:
        # Check if this is a heading
        if para.style.name.startswith("Heading"):
            # Add the heading to the new document
            new_para = new_doc.add_paragraph()
            new_para.style = new_doc.styles[f"Heading {para.style.name[-1]}"]
            new_para.add_run(para.text)
            found_content = True

        # Check for highlighted text in runs
        for run in para.runs:
            if run.font.highlight_color:
                # Add a new paragraph if needed
                if not found_content:
                    new_para = new_doc.add_paragraph()
                    found_content = True

                # Add highlighted text
                highlight_run = new_para.add_run(run.text)
                highlight_run.font.highlight_color = WD_COLOR_INDEX.YELLOW

    # Only save if we found content
    if found_content:
        new_doc.save(output_path)
        print(f"Extracted headings and highlights saved to {output_path}")
    else:
        print("No headings or highlighted text found in the document")


# Example usage
extract_headings_and_highlights("input.docx", "output.docx")
```

```python
import docx
from docx.enum.text import WD_COLOR_INDEX
from docx.shared import Pt


def extract_headings_and_highlights(input_path, output_path):
    # Open the input document
    doc = docx.Document(input_path)

    # Create a new document for output
    new_doc = docx.Document()

    # Track if we've found any content
    found_content = False

    # Iterate through paragraphs in the input document
    for para in doc.paragraphs:
        # Check if this is a heading
        if para.style.name.startswith("Heading"):
            # Add the heading to the new document
            new_para = new_doc.add_paragraph()
            new_para.style = new_doc.styles[f"Heading {para.style.name[-1]}"]

            # Add text with proper spacing
            for run in para.runs:
                if run.text.strip():  # Only add non-empty text
                    if new_para.text:  # If paragraph already has text, add space
                        new_para.add_run(" ")
                    new_para.add_run(run.text)
            found_content = True

        # Check for highlighted text in runs
        for run in para.runs:
            if run.font.highlight_color:
                # Add a new paragraph if needed
                if not found_content:
                    new_para = new_doc.add_paragraph()
                    found_content = True

                # Add highlighted text with proper spacing
                if new_para.text:  # If paragraph already has text, add space
                    new_para.add_run(" ")
                highlight_run = new_para.add_run(run.text)
                highlight_run.font.highlight_color = WD_COLOR_INDEX.YELLOW

    # Only save if we found content
    if found_content:
        # Set paragraph spacing
        for para in new_doc.paragraphs:
            para_format = para.paragraph_format
            para_format.space_before = Pt(12)  # 12pt space before
            para_format.space_after = Pt(0)  # No space after

        new_doc.save(output_path)
        print(f"Extracted headings and highlights saved to {output_path}")
    else:
        print("No headings or highlighted text found in the document")
```

```python
extract_headings_and_highlights("input.docx", "output.docx")
```
